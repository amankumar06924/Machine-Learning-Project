```python
import os
from langgraph.graph import StateGraph, START, END
from typing import TypedDict
from dotenv import load_dotenv
import google.genai as genai
```

```python
load_dotenv()
api_key = os.environ.get("GOOGLE_API_KEY")
if not api_key:
    raise RuntimeError("Please set GOOGLE_API_KEY in your environment or .env file.")
client = genai.Client(api_key=api_key)
```

```python
class Blogstate(TypedDict):
    title:str
    outline:str
    content:str
```

```python
def topic(state:Blogstate)->Blogstate:
    tit = state['title']
    prompt = f'Generate a outline for the topic - {tit}'
    response = client.models.generate_content(
        model='gemini-2.5-flash',
        contents=prompt,
    )
    ans = response.text

    state['outline'] = ans

    return state
```

```python
def blog(state:Blogstate)->Blogstate:
    out = state['outline']
    tit = state['title']
    prompt = f'write a detail blog on the title - {tit} using the following outline \n {out}'
    response = client.models.generate_content(
        model='gemini-2.5-flash',
        contents=prompt,
    )
    ans = response.text
    state['content'] = ans
    return state
```

```python
graph = StateGraph(Blogstate)

# node 
graph.add_node('create_topic',topic)
graph.add_node('create_blog',blog)
# edge
graph.add_edge(START,'create_topic')
graph.add_edge('create_topic','create_blog')
graph.add_edge('create_topic',END)

workflow = graph.compile()

```

```python
inital_state = {'title':'rise of ai in india'}
final_state = workflow.invoke(inital_state)

print(final_state['content'])
```
